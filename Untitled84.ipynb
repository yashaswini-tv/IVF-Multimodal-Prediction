```python
import pandas as pd

df = pd.read_excel("/content/Embryo dataset_2542.xlsx")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
# Create patient ID from embryo ID
df["patient_id"] = df["胚胎"].astype(str).str.split("_").str[0]

print("Embryo records:", len(df))
print("Unique patients:", df["patient_id"].nunique())

# Number of embryos per patient
print("\nEmbryos per patient:")
print(df.groupby("patient_id").size().value_counts().sort_index())

# Check whether any patient has conflicting pregnancy labels
label_check = df.groupby("patient_id")["label"].nunique()

print("\nPatients with conflicting labels:",
      (label_check > 1).sum())
```

```python
patient_labels = df.groupby("patient_id")["label"].first()

print("Total patients:", len(patient_labels))
print("\nPregnancy outcome:")
print(patient_labels.value_counts())

print("\nPercentage:")
print((patient_labels.value_counts(normalize=True) * 100).round(2))
```

```python
from sklearn.model_selection import train_test_split

# One row per patient for splitting
patients = (
    df.groupby("patient_id")["label"]
      .first()
      .reset_index()
)

# 70% training, 30% temporary
train_patients, temp_patients = train_test_split(
    patients,
    test_size=0.30,
    stratify=patients["label"],
    random_state=42
)

# Divide remaining 30% into 15% validation + 15% test
val_patients, test_patients = train_test_split(
    temp_patients,
    test_size=0.50,
    stratify=temp_patients["label"],
    random_state=42
)

train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

print("Patients")
print("Train:", len(train_ids))
print("Validation:", len(val_ids))
print("Test:", len(test_ids))

print("\nLeakage checks")
print("Train-Val overlap:", len(train_ids & val_ids))
print("Train-Test overlap:", len(train_ids & test_ids))
print("Val-Test overlap:", len(val_ids & test_ids))

print("\nPregnancy rate")
print("Train:", round(train_patients["label"].mean(), 3))
print("Validation:", round(val_patients["label"].mean(), 3))
print("Test:", round(test_patients["label"].mean(), 3))
```

```python
# Assign split to every embryo record
df["split"] = df["patient_id"].map(
    lambda x: "train" if x in train_ids
    else "validation" if x in val_ids
    else "test"
)

# Check embryo counts
print(df["split"].value_counts())

# Save the complete dataset with patient IDs and split
df.to_csv(
    "/content/IVF_dataset_patient_split.csv",
    index=False
)

print("\nSaved successfully!")
```

```python
import pandas as pd

# Continuous variables only
continuous_cols = [
    c for c in feature_cols
    if c != "男方精液处理方法"
]

outlier_summary = []

for col in continuous_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 3 * iqr
    upper = q3 + 3 * iqr

    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append([
        col,
        n_outliers,
        df[col].min(),
        df[col].max()
    ])

outlier_table = pd.DataFrame(
    outlier_summary,
    columns=["Variable", "Extreme_values", "Minimum", "Maximum"]
)

outlier_table = outlier_table.sort_values(
    "Extreme_values",
    ascending=False
)

display(outlier_table)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    accuracy_score,
    recall_score,
    precision_score,
    f1_score,
    confusion_matrix
)

# Build logistic regression model
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Train on training patients only
logistic_model.fit(X_train, y_train)

# Predict validation patients
val_probability = logistic_model.predict_proba(X_val)[:, 1]
val_prediction = (val_probability >= 0.5).astype(int)

# Calculate performance
auc = roc_auc_score(y_val, val_probability)
accuracy = accuracy_score(y_val, val_prediction)
sensitivity = recall_score(y_val, val_prediction)
specificity = recall_score(y_val, val_prediction, pos_label=0)
precision = precision_score(y_val, val_prediction, zero_division=0)
f1 = f1_score(y_val, val_prediction)

print("LOGISTIC REGRESSION — VALIDATION")
print("--------------------------------")
print("AUROC:", round(auc, 3))
print("Accuracy:", round(accuracy, 3))
print("Sensitivity:", round(sensitivity, 3))
print("Specificity:", round(specificity, 3))
print("Precision:", round(precision, 3))
print("F1 score:", round(f1, 3))

print("\nConfusion matrix:")
print(confusion_matrix(y_val, val_prediction))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Train on training patients only
rf_model.fit(X_train, y_train)

# Validation predictions
rf_probability = rf_model.predict_proba(X_val)[:, 1]
rf_prediction = (rf_probability >= 0.5).astype(int)

# Performance
rf_auc = roc_auc_score(y_val, rf_probability)
rf_accuracy = accuracy_score(y_val, rf_prediction)
rf_sensitivity = recall_score(y_val, rf_prediction)
rf_specificity = recall_score(y_val, rf_prediction, pos_label=0)
rf_precision = precision_score(y_val, rf_prediction, zero_division=0)
rf_f1 = f1_score(y_val, rf_prediction)

print("RANDOM FOREST — VALIDATION")
print("--------------------------")
print("AUROC:", round(rf_auc, 3))
print("Accuracy:", round(rf_accuracy, 3))
print("Sensitivity:", round(rf_sensitivity, 3))
print("Specificity:", round(rf_specificity, 3))
print("Precision:", round(rf_precision, 3))
print("F1 score:", round(rf_f1, 3))

print("\nConfusion matrix:")
print(confusion_matrix(y_val, rf_prediction))
```

```python
import pandas as pd
import os
import zipfile
from sklearn.model_selection import train_test_split

# File paths in Google Drive
excel_path = "/content/drive/MyDrive/Embryo dataset_2542.xlsx"
zip_path = "/content/drive/MyDrive/Data set splitting.zip"

# Load Excel
df = pd.read_excel(excel_path)

# Patient ID
df["patient_id"] = df["胚胎"].astype(str).str.split("_").str[0]

# Recreate EXACT same patient split
patients = (
    df.groupby("patient_id")["label"]
      .first()
      .reset_index()
)

train_patients, temp_patients = train_test_split(
    patients,
    test_size=0.30,
    stratify=patients["label"],
    random_state=42
)

val_patients, test_patients = train_test_split(
    temp_patients,
    test_size=0.50,
    stratify=temp_patients["label"],
    random_state=42
)

train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

df["split"] = df["patient_id"].map(
    lambda x: "train" if x in train_ids
    else "validation" if x in val_ids
    else "test"
)

# Unzip images
extract_path = "/content/embryo_images"

if not os.path.exists(extract_path):
    os.makedirs(extract_path)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_path)

print("Dataset restored!")
print("Rows:", len(df))
print("\nPatients:")
print(df.groupby("split")["patient_id"].nunique())

print("\nEmbryo images expected:")
print(df["split"].value_counts())
```

```python
# Find all embryo images and link them to the dataset

image_records = []

for root, dirs, files in os.walk("/content/embryo_images"):
    for file in files:
        if file.lower().endswith((".jpg", ".jpeg", ".png")):
            embryo_id = os.path.splitext(file)[0]

            image_records.append({
                "胚胎": embryo_id,
                "image_path": os.path.join(root, file)
            })

image_df = pd.DataFrame(image_records)

image_df = image_df.merge(
    df[["胚胎", "patient_id", "label", "split"]],
    on="胚胎",
    how="left"
)

print("Images found:", len(image_df))
print("Missing labels:", image_df["label"].isna().sum())

print("\nImages by split:")
print(image_df["split"].value_counts())
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_img_df = image_df[image_df["split"] == "train"].copy()
val_img_df = image_df[image_df["split"] == "validation"].copy()

def load_image(path, label):
    image = tf.io.read_file(path)
    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    image = tf.cast(image, tf.float32)

    return image, tf.cast(label, tf.float32)

train_ds = tf.data.Dataset.from_tensor_slices(
    (
        train_img_df["image_path"].values,
        train_img_df["label"].values
    )
)

train_ds = (
    train_ds
    .shuffle(
        len(train_img_df),
        seed=42,
        reshuffle_each_iteration=True
    )
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

val_ds = tf.data.Dataset.from_tensor_slices(
    (
        val_img_df["image_path"].values,
        val_img_df["label"].values
    )
)

val_ds = (
    val_ds
    .map(load_image, num_parallel_calls=tf.data.AUTOTUNE)
    .batch(BATCH_SIZE)
    .prefetch(tf.data.AUTOTUNE)
)

print("Training images:", len(train_img_df))
print("Validation images:", len(val_img_df))

for images, labels in train_ds.take(1):
    print("Image batch shape:", images.shape)
    print("Label batch shape:", labels.shape)
```

```python
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB0

# Pretrained EfficientNet
base_model = EfficientNetB0(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

# Freeze pretrained layers initially
base_model.trainable = False

# Build classifier
inputs = layers.Input(shape=(224, 224, 3))

# Light data augmentation — training only
x = layers.RandomFlip("horizontal")(inputs)
x = layers.RandomRotation(0.05)(x)
x = layers.RandomZoom(0.05)(x)

x = base_model(x, training=False)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.30)(x)

outputs = layers.Dense(
    1,
    activation="sigmoid"
)(x)

image_model = models.Model(inputs, outputs)

image_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.BinaryAccuracy(name="accuracy")
    ]
)

image_model.summary()

print("\nBase model trainable:",
      base_model.trainable)
```

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_auc",
        mode="max",
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]

history = image_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=callbacks
)
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score

# Predict each validation image
val_predictions = image_model.predict(
    val_ds,
    verbose=1
).ravel()

# Attach predictions to validation image table
val_results = val_img_df[
    ["patient_id", "label"]
].copy()

val_results["image_probability"] = val_predictions

# Average probabilities for patients with multiple embryo images
patient_val_results = (
    val_results
    .groupby("patient_id")
    .agg(
        label=("label", "first"),
        image_probability=("image_probability", "mean"),
        n_images=("image_probability", "size")
    )
    .reset_index()
)

patient_auc = roc_auc_score(
    patient_val_results["label"],
    patient_val_results["image_probability"]
)

print("Validation images:", len(val_results))
print("Validation patients:", len(patient_val_results))
print("Patient-level image AUROC:", round(patient_auc, 3))

print("\nImages per patient:")
print(patient_val_results["n_images"].value_counts().sort_index())
```

```python
# Unfreeze EfficientNet
base_model.trainable = True

# Keep most layers frozen
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep BatchNormalization layers frozen for stability
for layer in base_model.layers:
    if isinstance(layer, tf.keras.layers.BatchNormalization):
        layer.trainable = False

# Recompile with a much smaller learning rate
image_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="binary_crossentropy",
    metrics=[
        tf.keras.metrics.AUC(name="auc"),
        tf.keras.metrics.BinaryAccuracy(name="accuracy")
    ]
)

print(
    "Trainable parameters:",
    sum(
        np.prod(v.shape)
        for v in image_model.trainable_weights
    )
)
```

```python
fine_tune_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=5,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_auc",
        mode="max",
        factor=0.5,
        patience=2,
        min_lr=1e-7
    )
]

fine_history = image_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    callbacks=fine_tune_callbacks
)
```

```python
from sklearn.metrics import roc_auc_score
import numpy as np

def bootstrap_auc_ci(y_true, y_prob, n_bootstraps=2000, seed=42):
    rng = np.random.default_rng(seed)
    aucs = []

    y_true = np.asarray(y_true)
    y_prob = np.asarray(y_prob)

    for _ in range(n_bootstraps):
        idx = rng.choice(
            len(y_true),
            size=len(y_true),
            replace=True
        )

        # AUC requires both outcome classes
        if len(np.unique(y_true[idx])) < 2:
            continue

        aucs.append(
            roc_auc_score(
                y_true[idx],
                y_prob[idx]
            )
        )

    lower = np.percentile(aucs, 2.5)
    upper = np.percentile(aucs, 97.5)

    return lower, upper


y = final_test["label"].values

for name, column in [
    ("Clinical", "clinical_prob"),
    ("Image", "image_probability"),
    ("Multimodal", "multimodal_prob")
]:
    auc = roc_auc_score(y, final_test[column])

    lower, upper = bootstrap_auc_ci(
        y,
        final_test[column]
    )

    print(
        f"{name}: AUROC {auc:.3f} "
        f"(95% CI {lower:.3f}-{upper:.3f})"
    )
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def classification_metrics(y_true, probabilities, threshold=0.50):
    predictions = (np.asarray(probabilities) >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_true, predictions
    ).ravel()

    return {
        "Accuracy": accuracy_score(y_true, predictions),
        "Sensitivity": recall_score(y_true, predictions),
        "Specificity": tn / (tn + fp),
        "Precision": precision_score(
            y_true, predictions, zero_division=0
        ),
        "F1": f1_score(y_true, predictions),
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }

for name, column in [
    ("Clinical", "clinical_prob"),
    ("Image", "image_probability"),
    ("Multimodal", "multimodal_prob")
]:
    m = classification_metrics(
        final_test["label"],
        final_test[column]
    )

    print(f"\n{name}")
    print(
        f"Accuracy: {m['Accuracy']:.3f} | "
        f"Sensitivity: {m['Sensitivity']:.3f} | "
        f"Specificity: {m['Specificity']:.3f} | "
        f"Precision: {m['Precision']:.3f} | "
        f"F1: {m['F1']:.3f}"
    )
    print(
        f"TN={m['TN']}, FP={m['FP']}, "
        f"FN={m['FN']}, TP={m['TP']}"
    )
```

```python
import numpy as np
import statsmodels.api as sm

def calibration_intercept_slope(y_true, probabilities):
    p = np.clip(
        np.asarray(probabilities),
        1e-6,
        1 - 1e-6
    )

    logit_p = np.log(p / (1 - p))

    # Calibration intercept:
    # offset fixes the prediction coefficient at 1
    intercept_model = sm.GLM(
        y_true,
        np.ones((len(y_true), 1)),
        family=sm.families.Binomial(),
        offset=logit_p
    ).fit()

    intercept = intercept_model.params[0]

    # Calibration slope
    X = sm.add_constant(logit_p)

    slope_model = sm.GLM(
        y_true,
        X,
        family=sm.families.Binomial()
    ).fit()

    slope = slope_model.params[1]

    return intercept, slope


for name, column in [
    ("Clinical", "clinical_prob"),
    ("Image", "image_probability"),
    ("Multimodal", "multimodal_prob")
]:
    intercept, slope = calibration_intercept_slope(
        final_test["label"].values,
        final_test[column].values
    )

    print(
        f"{name}: "
        f"Calibration intercept = {intercept:.3f}, "
        f"slope = {slope:.3f}"
    )
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

excel_path = "/content/drive/MyDrive/Embryo dataset_2542.xlsx"

df = pd.read_excel(excel_path)

df["patient_id"] = (
    df["胚胎"].astype(str).str.split("_").str[0]
)

# Recreate the exact same patient split
patients = (
    df.groupby("patient_id")["label"]
    .first()
    .reset_index()
)

train_patients, temp_patients = train_test_split(
    patients,
    test_size=0.30,
    stratify=patients["label"],
    random_state=42
)

val_patients, test_patients = train_test_split(
    temp_patients,
    test_size=0.50,
    stratify=temp_patients["label"],
    random_state=42
)

train_ids = set(train_patients["patient_id"])
val_ids = set(val_patients["patient_id"])
test_ids = set(test_patients["patient_id"])

df["split"] = df["patient_id"].map(
    lambda x: "train" if x in train_ids
    else "validation" if x in val_ids
    else "test"
)

feature_cols = [
    col for col in df.columns
    if col not in ["胚胎", "label", "patient_id", "split"]
]

clinical_df = (
    df.drop_duplicates("patient_id")
    [["patient_id", "label", "split"] + feature_cols]
    .copy()
)

val_df = clinical_df[
    clinical_df["split"] == "validation"
].copy()

X_val = val_df[feature_cols]
y_val = val_df["label"]

print("Clinical predictors:", len(feature_cols))
print("Validation patients:", len(val_df))
```

```python
from sklearn.inspection import permutation_importance
import pandas as pd
import matplotlib.pyplot as plt

perm = permutation_importance(
    clinical_model,
    X_val,
    y_val,
    scoring="roc_auc",
    n_repeats=30,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": feature_cols,
    "Importance": perm.importances_mean,
    "SD": perm.importances_std
}).sort_values("Importance", ascending=False)

top15 = importance_df.head(15).sort_values(
    "Importance",
    ascending=True
)

plt.figure(figsize=(8, 7))

plt.barh(
    top15["Feature"],
    top15["Importance"],
    xerr=top15["SD"],
    capsize=3
)

plt.xlabel("Decrease in Validation AUROC")
plt.ylabel("Clinical Feature")
plt.title("Permutation Importance of Clinical Predictors")
plt.tight_layout()

plt.savefig(
    "/content/drive/MyDrive/IVF_clinical_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Feature-importance figure saved to Google Drive.")
```

```python
def make_gradcam_heatmap(img_batch, image_model, base_model):

    # Pass through augmentation layers first
    x = img_batch
    for layer in image_model.layers[1:4]:
        x = layer(x, training=False)

    # Get EfficientNet feature maps
    with tf.GradientTape() as tape:
        conv_output = base_model(x, training=False)
        tape.watch(conv_output)

        # Pass through the classifier head
        y = image_model.get_layer(
            "global_average_pooling2d"
        )(conv_output)

        y = image_model.get_layer("dropout")(
            y, training=False
        )

        prediction = image_model.get_layer("dense")(y)

        class_score = prediction[:, 0]

    # Gradient of prediction with respect to feature maps
    grads = tape.gradient(
        class_score,
        conv_output
    )

    # Average gradient for each channel
    weights = tf.reduce_mean(
        grads,
        axis=(1, 2)
    )

    # Weighted combination of feature maps
    heatmap = tf.reduce_sum(
        conv_output * weights[:, None, None, :],
        axis=-1
    )

    # Keep positive contributions
    heatmap = tf.maximum(heatmap, 0)

    # Normalize to 0–1
    max_value = tf.reduce_max(heatmap)

    heatmap = tf.where(
        max_value > 0,
        heatmap / max_value,
        heatmap
    )

    return heatmap[0].numpy()


print("Grad-CAM function created successfully.")
```

```python
# Generate Grad-CAM heatmap
heatmap = make_gradcam_heatmap(
    img_batch,
    image_model,
    base_model
)

print("Heatmap shape:", heatmap.shape)
print("Minimum value:", round(float(heatmap.min()), 4))
print("Maximum value:", round(float(heatmap.max()), 4))

# Display the heatmap
plt.figure(figsize=(5, 5))
plt.imshow(heatmap, cmap="jet")
plt.colorbar(label="Grad-CAM activation")
plt.title(
    f"Grad-CAM: Embryo {sample_row['胚胎']}\n"
    f"True label={int(sample_row['label'])}, "
    f"Predicted probability={prediction:.3f}"
)
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))

for ax, (_, row) in zip(axes.flat, selected_examples.iterrows()):

    # Load image
    img_temp = tf.keras.utils.load_img(
        row["image_path"],
        target_size=(224, 224)
    )

    img_array_temp = tf.keras.utils.img_to_array(img_temp)
    img_batch_temp = tf.expand_dims(img_array_temp, axis=0)

    # Grad-CAM
    hm = make_gradcam_heatmap(
        img_batch_temp,
        image_model,
        base_model
    )

    hm_resized = tf.image.resize(
        hm[..., np.newaxis],
        (224, 224)
    ).numpy().squeeze()

    # Display original + heatmap
    ax.imshow(np.array(img_temp).astype("uint8"))
    ax.imshow(
        hm_resized,
        cmap="jet",
        alpha=0.40,
        vmin=0,
        vmax=1
    )

    ax.set_title(
        f"Embryo {row['胚胎']}\n"
        f"True label={int(row['label'])}, "
        f"Predicted={row['predicted_probability']:.3f}"
    )

    ax.axis("off")

plt.suptitle(
    "Grad-CAM Visualization of Validation Images",
    fontsize=15
)

plt.tight_layout()

save_path = "/content/drive/MyDrive/IVF_GradCAM_validation_examples.png"

plt.savefig(
    save_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved to:", save_path)
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score

y_true = final_preds["label"].values
clinical_prob = final_preds["clinical_prob"].values
multimodal_prob = final_preds["multimodal_prob"].values

# Observed AUROCs
clinical_auc = roc_auc_score(y_true, clinical_prob)
multimodal_auc = roc_auc_score(y_true, multimodal_prob)

observed_difference = multimodal_auc - clinical_auc

# Paired patient-level bootstrap
rng = np.random.default_rng(42)
n_bootstrap = 2000
differences = []

n = len(y_true)

for _ in range(n_bootstrap):
    idx = rng.choice(n, size=n, replace=True)

    y_boot = y_true[idx]

    # Skip rare bootstrap samples containing only one class
    if len(np.unique(y_boot)) < 2:
        continue

    clinical_boot = roc_auc_score(
        y_boot,
        clinical_prob[idx]
    )

    multimodal_boot = roc_auc_score(
        y_boot,
        multimodal_prob[idx]
    )

    differences.append(
        multimodal_boot - clinical_boot
    )

lower = np.percentile(differences, 2.5)
upper = np.percentile(differences, 97.5)

print("Clinical AUROC:", round(clinical_auc, 4))
print("Multimodal AUROC:", round(multimodal_auc, 4))
print(
    "AUROC difference (multimodal - clinical):",
    round(observed_difference, 4)
)
print(
    "95% bootstrap CI for difference:",
    f"{lower:.4f} to {upper:.4f}"
)
```

```python
auc_comparison = pd.DataFrame({
    "clinical_auc": [clinical_auc],
    "multimodal_auc": [multimodal_auc],
    "auc_difference": [observed_difference],
    "difference_ci_lower": [lower],
    "difference_ci_upper": [upper],
    "bootstrap_repetitions": [2000]
})

auc_comparison.to_csv(
    "/content/drive/MyDrive/IVF_clinical_vs_multimodal_auc_difference.csv",
    index=False
)

print("AUROC comparison saved successfully.")
```
